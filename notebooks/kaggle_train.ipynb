```python
import yaml
import os

# Create dataset.yaml for Kaggle
dataset_config = {
    'path': '/kaggle/working/data/processed',
    'train': 'images/train',
    'val': 'images/val',
    'test': 'images/test',
    'nc': 6,
    'names': {
        0: 'crazing',
        1: 'inclusion',
        2: 'patches',
        3: 'pitted_surface',
        4: 'rolled-in_scale',
        5: 'scratches'
    }
}

os.makedirs('/kaggle/working/configs', exist_ok=True)
with open('/kaggle/working/configs/dataset.yaml', 'w') as f:
    yaml.dump(dataset_config, f)
    
print('Created dataset.yaml')
```

```python
from ultralytics import YOLO

# Load a pretrained YOLO model
model = YOLO('yolov8s.pt')

# Train the model
results = model.train(
    data='/kaggle/working/configs/dataset.yaml',
    epochs=100,
    patience=20,
    batch=16,
    imgsz=640,
    device='0', # 0 for GPU
    project='/kaggle/working/runs/train',
    name='smart_defect_detector_v1',
    save=True,
    plots=True
)
```
